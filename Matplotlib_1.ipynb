```python
batsman = pd.read_csv('/content/sharma-kohli.csv')
```

```python
batsman
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
```

```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
# using tehn hidden parameters
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black',linestyle='dashed',marker='D')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
# legend -> location
# using tehn hidden parameters
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black',linestyle='dashed',marker='D',label='rohit')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')

plt.legend(loc = 'best') #if u cant call this fun this legends will not gonna appear
```

```python
#limiting axes  #trimming-> if there is any outlier value comes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,100000)
plt.xlim(2017,2019)
```

```python
#grid

plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')

plt.grid()
plt.show()
```
