```python
# some python course
s = ['uttam','kantit','soumyasen']
def name_len(l:list):
    len_l = []
    for i in l:
        len_l.append(len(i))
    d = dict(zip(s,len_l))
    for i in l:
        if len(i) == max(d.values()):
            return i
print(name_len(s))
```
