```python
# if we already have datetime object in python then why pandas makes timestamp?
```

- syntax wise datetime is very convenient
- But the performance takes a hit while working with huge data. List vs Numpy Array
- The weaknesses of Python's datetime format inspired the NumPy team to add a set of native time series data type to NumPy.
- The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.

```python
import numpy as np
date = np.array('2016-09-22', dtype=np.datetime64)
date
```

```python
# vectorized operations on it

date + np.arange(12)
```

```python
# full story python is slow,
# thats why the pandas and numpy guys make their own datetime object as timestamp
```

- Because of the uniform type in NumPy datetime64 arrays, this type of operation can be accomplished much more quickly than if we were working directly with Python's datetime objects, especially as arrays get large

- Pandas Timestamp object combines the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64

- From a group of these Timestamp objects, Pandas can construct a DatetimeIndex that can be used to index data in a Series or DataFrame

```python
# to store the multiple timestamps what to do?  DatetimeIndex Object
```

### DatetimeIndex Object

A collection of pandas timestamp

```python
pd.DatetimeIndex(['2023/1/1','2021/1/1','2023/5/4'])
```

```python
# using python datetime object
pd.DatetimeIndex([dt.datetime(2023,1,1),dt.datetime(2022,1,1),dt.datetime(2021,1,1)])
```

```python
# using pd.timestamps
dt_index = pd.DatetimeIndex([pd.Timestamp(2023,1,1),pd.Timestamp(2022,1,1),pd.Timestamp(2021,1,1)])
```

```python
pd.Series([1,2,3],index=dt_index)
```

```python
# Date_range functions
```

```python
pd.date_range(start='2023/1/1',end='2023/1/18',freq='D')
```

```python
pd.date_range(start='2023/1/1',end='2023/1/18',freq='3D')
```

```python
# B -> business days
pd.date_range(start='2023/1/5',end='2023/2/28',freq='B')
```

```python
# W -> one week per day
pd.date_range(start='2023/1/5',end='2023/2/28',freq='W-THU')
```

```python
# H -> Hourly data(factor)
pd.date_range(start='2023/1/5',end='2023/2/28',freq='6H')
```

```python
# M -> Month end
pd.date_range(start='2023/1/5',end='2023/2/28',freq='M')
```

```python
# MS -> Month start
pd.date_range(start='2023/1/5',end='2023/2/28',freq='MS')
```

```python
# A -> Year end
pd.date_range(start='2023/1/5',end='2030/2/28',freq='A')
```

```python
# using periods(number of results)
pd.date_range(start='2023/1/5',periods=25,freq='M')
```

### to_datetime function

converts an existing objects to pandas timestamp/datetimeindex object

```python
#simple series example

s = pd.Series(['2023/1/1','2021/1/1','2022/1/1'])
s
```

```python
s.str.split('/').str.get(0)
```

```python
# series-> to_datetme

#here dt is the accesor here

pd.to_datetime(s)
```

```python
pd.to_datetime(s).dt.year
```
