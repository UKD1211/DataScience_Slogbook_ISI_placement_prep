```python
import seaborn as sns

# Load Titanic dataset
titanic = sns.load_dataset("titanic")

# Show first 5 rows
print(titanic.head())
```

```python
titanic.shape[0]
```

```python
titanic['age'].mean()
```

```python
sns.kdeplot(titanic['age'])
```

```python
titanic['age'].std()
```

```python
# now if i do the stadardization here

titanic['standard_age'] = (titanic['age'] - titanic['age'].mean())/titanic['age'].std()
```

```python
titanic.head()
```

```python
sns.kdeplot(titanic['standard_age'])
```

```python
titanic['age'].skew()
```

```python
titanic['age'].mean()+3*titanic['age'].std()
```
