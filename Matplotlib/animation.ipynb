```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# Create figure and axis
fig, axis = plt.subplots()
axis.set_xlim(-10, 10)
axis.set_ylim(-2, 2)

# Empty line to be updated
animated_plot, = axis.plot([], [], lw=2)
x = np.linspace(-10, 10, 400)

# Update function
def update_data(frame):
    y = np.tan(x + frame * 0.1)
    animated_plot.set_data(x, y)
    return animated_plot,

# Create the animation
animation = FuncAnimation(fig, update_data, frames=200, interval=50, blit=True)

# Display the animation in Colab
HTML(animation.to_jshtml())
```
