```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
# labels title
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
# using tehn hidden parameters
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black',linestyle='dashed',marker='D')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')
```

```python
# legend -> location
# using tehn hidden parameters
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed',linewidth=3,marker='D',label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black',linestyle='dashed',marker='D',label='rohit')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')

plt.legend(loc = 'best') #if u cant call this fun this legends will not gonna appear
```

```python
#limiting axes  #trimming-> if there is any outlier value comes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,100000)
plt.xlim(2017,2019)
```

```python
#grid

plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'], color='black')

plt.title('Rohit Sharma vs V kohli')
plt.xlabel('Year')
plt.ylabel('runs scored')

plt.grid()
plt.show()
```

### Bar chart

image.png

- Bivariate Analysis
- Numerical vs Categorical
- Use case - Aggregate analysis of groups

```python
children = [10,15,20,35,56]
colors = ['red','blue','green','yellow','pink']
plt.bar(colors,children,color='black')
```

```python
#horizontal bar chart
plt.barh(colors,children,color='black')
```

```python
#colors and label
df = pd.read_csv('/content/batsman_season_record.csv')
```

```python
df
```

```python
plt.bar(df['batsman'],df['2015'],width=0.5)
```

```python
plt.bar(np.arange(df.shape[0])-0.2,df['2015'],width=0.2)
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2)
plt.bar(np.arange(df.shape[0])+0.2,df['2017'],width=0.2)

plt.xticks(np.arange(df.shape[0]),df['batsman'])
plt.show()
```

```python
# overlapping in the xaxis
children = [10,15,20,35,56]
colors = ['red red red red','blue blue blue blue','green green green','yellow yellow yellow','pink pink']
plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
plt.show()
```

```python
#stacked bar chart
df
```

```python
plt.bar(df['batsman'],df['2015'])
plt.bar(df['batsman'],df['2016'],bottom=df['2015'])
plt.bar(df['batsman'],df['2017'],bottom=df['2016']+df['2015'])
```

### Pie Chart

image.png

- Univariate/Bivariate Analysis
- Categorical vs numerical
- Use case - To find contibution on a standard scale

```python
#simple data

data = [23,45,100,20,80,49]
subjects = ['eng','maths','science','social','hindi','history']
plt.pie(data, labels = subjects)
plt.show()
```

```python
#dataset
df = pd.read_csv('/content/gayle-175.csv')
```

```python
df
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'],autopct="%0.1f%%")
plt.show()
```

```python
plt.pie(df['batsman_runs'], labels=df['batsman'],autopct="%0.1f%%", colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
#explode shadow
plt.pie(df['batsman_runs'], labels=df['batsman'],autopct="%0.1f%%", explode=[0.3,0,0,0,0,0.1], shadow=True )
plt.show()
```
