```python
iris = pd.read_csv('iris.csv')
```

```python
iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
plt.show()
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'],cmap='winter')
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
# plt.show()
plt.colorbar()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'],cmap='winter',alpha=0.9)
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
plt.colorbar()
```

```python
# here we have learn about the color encoding on A 2D scatter plot
```

# **Surface Plot**

```python
#3D
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)
```

```python
x
```

```python
xx, yy = np.meshgrid(x,y)
```

```python
z = xx**2 + yy**2
```

```python
z.shape
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot(projection = '3d')
surface = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```

```python
z = np.sin(xx) + np.cos(yy)
fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot(projection = '3d')
surface = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```

```python
z = np.tan(xx) + np.tan(yy)
fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot(projection = '3d')
surface = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```

```python
z = np.sin(xx) + np.log(yy)
fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot(projection = '3d')
surface = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```

# **Contour Plot**

```python
# 3d graph to 2D representattion
```

```python
fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot()
surface = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```

```python
# filled contour graph

fig = plt.figure(figsize = (12,8))
ax = fig.add_subplot()
surface = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(surface)
```
