```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Colored Scatterplots
```

```python
iris = pd.read_csv('iris.csv')
```

```python
iris.head()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'])
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
plt.show()
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
iris.sample(5)
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'],cmap='winter')
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
# plt.show()
plt.colorbar()
```

```python
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'],cmap='winter',alpha=0.9)
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
plt.colorbar()
```

```python
# here we have learn about the color encoding on A 2D scatter plot
```

```python
# Plot Size
```

```python
plt.figure(figsize=(15,7))
plt.scatter(iris['SepalLengthCm'],iris['PetalLengthCm'], c=iris['Species'],cmap='winter')
plt.xlabel('Sepal_length')
plt.ylabel('Petal_length')
plt.colorbar()
```

```python
# Annotations
```

```python
batters = pd.read_csv('/content/batter (1).csv')
```

```python
batters
```

```python
sample_df = batters.head(100).sample(25,random_state=25)
```

```python
sample_df
```

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
# but from this we cant understand who is the which batsman actually , this problem is solved by the annotations
```

```python
x = [1,2,3,4]
y = [5,6,7,9]
```

```python
plt.scatter(x,y)
plt.text(1,5,'First')
plt.text(2,6,'Second')
plt.text(3,7,'Third')
plt.text(4,9,'4th')   #points labelling
```

# **Horizontal and Verical Line**

```python
plt.figure(figsize=(18,10))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs'])

plt.axhline(130,color='red')
plt.axhline(140,color='green')
plt.axvline(30,color='red')
for i in range(sample_df.shape[0]):
  plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i], sample_df['batter'].values[i])
```

```python
fig,ax = plt.subplots(nrows=2,ncols=1,sharex=True)
ax[0].scatter(batters['avg'],batters['strike_rate'],color='Red')
ax[1].scatter(batters['avg'],batters['runs'])
ax[0].set_title('Avg Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')


ax[1].set_title('Avg Vs Runs')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batters['avg'],batters['strike_rate'],color='Red')
ax[0,1].scatter(batters['avg'],batters['runs'])
ax[1,0].hist(batters['avg'])
ax[1,1].hist(batters['runs'])
```
