```python
import re

def remove_urls(text: str) -> str:
    """
    Removes URLs (http, https, www) from a given text string.
    """
    pattern = r'http[s]?://\S+|www\.\S+'
    return re.sub(pattern, '', text).strip()

# Example usage
sample = "Follow me at https://twitter.com/user and visit www.example.com"
print(remove_urls(sample))
```

```python
df['review'] = df['review'].apply(remove_urls)
```

```python
def chat_convert(text):      #good approach
  new_text = []
  for w in text.split():
    if w.lower() in chat_words:
      new_text.append(chat_words[w.lower()])
    else:
      new_text.append(w)
  return " ".join(new_text)
```

```python
def remove_stopwords(text):
  new_text = []
  for word in text.split():
    if word in stopwords.words('english'):
      new_text.append('')
    else:
      new_text.append(word)
  return " ".join(new_text)
```

```python
import re

def remove_emojis(text: str) -> str:
    """
    Removes emojis from a given text string.
    """
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags
        u"\U00002500-\U00002BEF"  # Chinese characters
        u"\U00002702-\U000027B0"
        u"\U00002702-\U000027B0"
        u"\U000024C2-\U0001F251"
        u"\U0001f926-\U0001f937"
        u"\U00010000-\U0010ffff"
        u"\u2640-\u2642"
        u"\u2600-\u2B55"
        u"\u200d"
        u"\u23cf"
        u"\u23e9"
        u"\u231a"
        u"\ufe0f"  # dingbats
        u"\u3030"
        "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)

# Example usage
sample = "Hello 😃🔥✨🌍! This is cool 🚀."
print(remove_emojis(sample))
```

```python
#2. Regular Expression
import re

def tokenize(text: str):
    """
    Tokenizes a given text string using regular expressions.
    Splits into words while ignoring punctuation.
    """
    # This regex matches words (alphanumeric + underscore)
    tokens = re.findall(r'\b\w+\b', text.lower())
    return tokens

# Example usage
sample = "Hello! This is a test, with numbers 123 and symbols #wow."
print(tokenize(sample))
```

```python
import nltk
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer, WordNetLemmatizer

# Download WordNet (only needed for lemmatization)
nltk.download("wordnet")
nltk.download("omw-1.4")

# Sample text
words = ["running", "flies", "better", "studies", "happily", "cars"]

# --- Stemming ---
porter = PorterStemmer()
snowball = SnowballStemmer("english")
lancaster = LancasterStemmer()

print("Stemming results:")
for w in words:
    print(f"{w:10} | Porter: {porter.stem(w)} | Snowball: {snowball.stem(w)} | Lancaster: {lancaster.stem(w)}")

# --- Lemmatization ---
lemmatizer = WordNetLemmatizer()

print("\nLemmatization results:")
for w in words:
    print(f"{w:10} | Lemma: {lemmatizer.lemmatize(w)}")
```
