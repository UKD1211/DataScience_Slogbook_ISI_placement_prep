```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
sns.kdeplot(df['sepal_length'])
```

```python
temp = sorted(df['sepal_length'].tolist())
```

```python
y_quant = []

for i in range(1,101):
  y_quant.append(np.percentile(temp,i))
```

```python
y_quant
```

```python
samples = np.random.normal(loc=0,scale=1,size=1000)  #sample normal data geenrates
```

```python
x_quant = []

for i in range(1,101):
  x_quant.append(np.percentile(samples,i))
```

```python
sns.scatterplot(x=x_quant,y=y_quant)
```

```python
plt.scatter(x_quant,y_quant)
```

```python
## using the statsmodel

import statsmodels.api as sm
import matplotlib.pyplot as plt

#create a QQ plot of the 2 sets of the data
fig = sm.qqplot(df['sepal_length'], line='45', fit=True)

plt.title('QQ plot')
plt.xlabel('Theoretical Quantiles')
plt.ylabel('sample Values')


plt.show()
```
